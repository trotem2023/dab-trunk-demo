```python
dbutils.widgets.text("catalog", "workspace")
dbutils.widgets.text("schema", "demo_dev")
dbutils.widgets.text("env", "dev")
	
catalog = dbutils.widgets.get("catalog")
schema  = dbutils.widgets.get("schema")
env     = dbutils.widgets.get("env")

from pyspark.sql import functions as F

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{schema}")

df = (spark.createDataFrame(
        [(1, "laptop", 4200.0), (2, "screen", 900.0), (3, "mouse", 120.0)],
        "order_id INT, product STRING, amount DOUBLE")
      .withColumn("env", F.lit(env))
      .withColumn("loaded_at", F.current_timestamp()))

df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable(f"{catalog}.{schema}.demo_orders")
display(spark.table(f"{catalog}.{schema}.demo_orders"))
```
